# Convo Capsule: Custom Speaker Diarization Model

This project builds a custom Deep Learning model for Speaker Diarization from scratch.

## Architecture
- **Feature Extraction**: Mel Filterbanks (80 features)
- **Encoder**: ECAPA-TDNN (Emphasized Channel Attention, Propagation and Aggregation)
- **Loss Function**: Angular Additive Margin Softmax (AAM-Softmax) (To be implemented for training)
- **Clustering**: Spectral Clustering

## Dependencies
`torch`, `torchaudio`, `numpy`, `scipy`, `sklearn`

```python
import torch
import numpy as np
from model import ECAPA_TDNN
from audio_processing import get_mel_spectrogram, segment_audio

print("Torch Version:", torch.__version__)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using Device:", device)
```

```python
# 1. Initialize the Model
model = ECAPA_TDNN(C=512).to(device)

# Determine total parameters
total_params = sum(p.numel() for p in model.parameters())
print(f"Total Trainable Parameters: {total_params:,}")
```

```python
# 2. Test with Dummy Input
# Simulate a Batch of Audio Spectrograms
# Shape: (Batch, n_mels, time)
# e.g., Batch=4, Mels=80, Time=200 (approx 2 seconds)

dummy_input = torch.randn(4, 80, 200).to(device)

model.eval() # Set to evaluation mode
with torch.no_grad():
    embedding = model(dummy_input)

print("Input Shape:", dummy_input.shape)
print("Output Embedding Shape:", embedding.shape)
print("Success! The model ingested audio features and produced speaker embeddings.")
```
